```python
import numpy as np
import pandas as pd

episode_labels_path = r"C:\Users\user\Datasets\hv_double_line_90kv\labels_filtered.csv"
# episode_labels_path = r"C:\Users\user\Datasets\hv_double_line_90kv_labels.csv"

episode_labels = pd.read_csv(episode_labels_path,sep=";")  # , sep=";", decimal=",")

columns_to_convert_to_numeric = [
    "t_evnt_start",
    "t_evnt_end",
    "sc_location",
    "fault_resistance",
    "line_1_2_a_length",
    "line_1_2_b_length",
    "line_2_3_a_length",
    "line_2_3_b_length",
    "line_1_2_a_xline",
    "line_1_2_a_rline",
    "line_1_2_a_cline",
    "line_1_2_a_xline0",
    "line_1_2_a_rline0",
    "line_1_2_a_cline0",
    "line_1_2_b_xline",
    "line_1_2_b_rline",
    "line_1_2_b_cline",
    "line_1_2_b_xline0",
    "line_1_2_b_rline0",
    "line_1_2_b_cline0",
    "line_2_3_a_xline",
    "line_2_3_a_rline",
    "line_2_3_a_cline",
    "line_2_3_a_xline0",
    "line_2_3_a_rline0",
    "line_2_3_a_cline0",
    "line_2_3_b_xline",
    "line_2_3_b_rline",
    "line_2_3_b_cline",
    "line_2_3_b_xline0",
    "line_2_3_b_rline0",
    "line_2_3_b_cline0",
]
for column in columns_to_convert_to_numeric:
    episode_labels[column] = pd.to_numeric(episode_labels[column], errors="coerce")

# Create t_event_duration column
episode_labels["t_event_duration"] = (
    episode_labels["t_evnt_end"] - episode_labels["t_evnt_start"]
)

episode_labels.info()
```

```python
# Build fault_class column
def classify_fault(row):
    if row["sc_type"] == 0: # 3 phase short-circuit
        return "flt_3ph_shc" + "_ABC"
    elif row["sc_type"] == 1: # 2 phase short-circuit
        if row["phase_select"] == 0:
            return "flt_2ph_shc_AB"
        elif row["phase_select"] == 1:
            return "flt_2ph_shc_BC"
        elif row["phase_select"] == 2:
            return "flt_2ph_shc_CA"
    elif row["sc_type"] == 2: # 1 phase short-circuit (always to ground)
        if row["phase_select"] == 0:
            return "flt_1phg_shc_A"
        elif row["phase_select"] == 1:
            return "flt_1phg_shc_B"
        elif row["phase_select"] == 2:
            return "flt_1phg_shc_C"
    elif row["sc_type"] == 3: # 2 phase to ground short-circuit
        if row["phase_select"] == 0:
            return "flt_2phg_shc_AB"
        elif row["phase_select"] == 1:
            return "flt_2phg_shc_BC"
        elif row["phase_select"] == 2:
            return "flt_2phg_shc_CA"
    else:
        raise ValueError("Unknown sc_type")
    
    

fault_class_list = episode_labels.apply(classify_fault, axis=1)
episode_labels["fault_class"] = fault_class_list
# move fault_class to be the first column
cols = episode_labels.columns.tolist()
cols = [cols[-1]] + cols[:-1]
episode_labels = episode_labels[cols]
```

```python
# Create LateX table with label counts of "fault_target", including percentages
label_counts = episode_labels['fault_target'].value_counts()
label_percentages = episode_labels['fault_target'].value_counts(normalize=True) * 100
label_summary = pd.DataFrame({'Count': label_counts, 'Percentage': label_percentages})
print(label_summary.to_latex(float_format="%.2f"))
print(f"Total number of fault targets: {label_counts.sum()}")
```

```python
def fault_type_from_fault_class(fc: str) -> str:
    fc = str(fc)
    if "flt_1phg" in fc:
        return "1ph-g"
    if "flt_2phg" in fc:
        return "2ph-g"
    if "flt_2ph" in fc:
        return "2ph"
    if "flt_3ph" in fc:
        return "3ph"
    raise ValueError(f"Unknown fault_class: {fc}")


episode_labels = episode_labels.copy()
episode_labels["fault_type"] = episode_labels["fault_class"].map(
    fault_type_from_fault_class
)
```

```python
LINE_MAP = {
    "Line_1_2_a": "1--2 (A)",
    "Line_1_2_b": "1--2 (B)",
    "Line_2_3_a": "2--3 (A)",
    "Line_2_3_b": "2--3 (B)",
}

LINE_ORDER = ["1--2 (A)", "1--2 (B)", "2--3 (A)", "2--3 (B)"]

episode_labels["fault_line"] = episode_labels["fault_target"].map(LINE_MAP)
```

```python
episode_labels
```

```python
ct = pd.crosstab(episode_labels["fault_type"], episode_labels["fault_line"])

ct = ct[LINE_ORDER]  # enforce physical ordering
ct["All"] = ct.sum(axis=1)

total_row = ct.sum(axis=0).to_frame().T
total_row.index = ["All"]
ct = pd.concat([ct, total_row])


core_rows = ct.index[:-1]
core_cols = ct.columns[:-1]

col_pct = ct.loc[core_rows, core_cols].div(ct.loc["All", core_cols], axis=1) * 100

out = ct.copy().astype(object)

for r in core_rows:
    for c in core_cols:
        n = int(ct.loc[r, c])
        p = col_pct.loc[r, c]
        out.loc[r, c] = f"{n:,} ({p:.1f}\\%)".replace(",", "{,}")

# margins = plain counts
for r in out.index:
    out.loc[r, "All"] = f"{int(ct.loc[r, 'All']):,}".replace(",", "{,}")
for c in out.columns:
    out.loc["All", c] = f"{int(ct.loc['All', c]):,}".replace(",", "{,}")


latex = out.to_latex(
    escape=False,
    column_format="lrrrrr",
    caption=(
        "High-level fault-type distribution per faulted line. "
        "Columns correspond to the four physical line sections; "
        "cells report the number of fault episodes and the share within each line."
    ),
    label="tab:faulttype_by_line",
)

print(latex)
```

```python
import numpy as np
import pandas as pd

path_to_labels = r"C:\Users\user\Repositories\fault-classification-localization\y_hv_double_line_90kv_W0p020_S0p005.parquet"

df_labels = pd.read_parquet(path_to_labels)

df_labels
```

```python
print(
    df_labels.loc[
        df_labels["status"] == "fault_start",
        ["event_type", "y_is_grounded", "y_phase_A", "y_phase_B", "y_phase_C"],
    ]
    .value_counts()
    .rename("count")
    .reset_index()
    .sort_values("count", ascending=False)
)
```

```python
def build_fault_label(row):
    phases = []
    if row["y_phase_A"]:
        phases.append("A")
    if row["y_phase_B"]:
        phases.append("B")
    if row["y_phase_C"]:
        phases.append("C")

    if phases:
        return f"{row['event_type']}_{''.join(phases)}"
    else:
        return row["event_type"]


df_labels["fault_label"] = df_labels.apply(build_fault_label, axis=1)

df_labels["fault_label"].value_counts()
```

```python
sample_ids = df_labels["sample_id"].to_numpy(dtype=int)
target_label = "event_type"

if target_label not in df_labels.columns:
    raise ValueError(
        f"Fault target column '{target_label}' not found in labels DataFrame. "
        f"Available columns: {df_labels.columns.tolist()}"
    )

y = df_labels[target_label]

fault_target_types = {
    "y_fault_present": int,
    "y_fault_location": float,
    "event_type": str,
}

if target_label not in fault_target_types:
    raise ValueError(
        f"Invalid target_label '{target_label}'. Supported labels: {list(fault_target_types.keys())}"
    )

y = y.astype(fault_target_types[target_label])
print(f"Extracted {len(sample_ids)} samples with fault target '{target_label}'.")
# print shape of y
print(f"Shape of target variable y: {y.shape}")

# sample_ids, np.array(y)

# print unique values in y and their counts
unique, counts = np.unique(y, return_counts=True)
print("Unique values in target variable y and their counts:")
for u, c in zip(unique, counts):
    print(f"Value: {u}, Count: {c}")
```
